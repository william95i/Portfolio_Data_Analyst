```python
#Combien mange en moyenne un être humain ? Source => ### Combien mange en moyenne un être humain ?

#Un adulte consomme en moyenne entre 2000 et 2500 kilocalories par jour, selon son sexe, son âge et son niveau d'activité physique
#Pour simplifier les calculs, nous utiliserons une valeur fixe de 2500 kcal/jour par personne.
#Cette valeur est stable entre 2017 et 2025, selon les recommandations nutritionnelles internationales.

#Source : [ANSES – Apports nutritionnels](https://www.anses.fr/fr/content/les-apports-nutritionnels-recommandés)
```

```python
#On commence par faire une jointure entre le data frame population et Dispo_alimentaire afin d'ajouter dans ce dernier la population
pop_2017 = population[population["Année"] == 2017]

dispo_pop = pd.merge(dispo_alimentaire, pop_2017, on="Zone", how="left")
```

```python
#Affichage du nouveau dataframe
print(dispo_pop.shape)   
dispo_pop.head(5)
```

```python
#Création de la colonne dispo_kcal avec calcul des kcal disponibles mondialement
dispo_pop["dispo_kcal"] = dispo_pop["Disponibilité alimentaire (Kcal/personne/jour)"] * dispo_pop["Population"]
# Total mondial de kilocalories disponibles
total_kcal = dispo_pop["dispo_kcal"].sum()
print(f"Total mondial de kilocalories disponibles : {total_kcal:,.0f} kcal/jour")
```

```python
#Calcul du nombre d'humains pouvant être nourris
# Apport moyen par personne
kcal_par_personne = 2500

# Nombre théorique de personnes pouvant être nourries
nb_personnes_nourries = total_kcal / kcal_par_personne

print(f"Nombre théorique de personnes pouvant être nourries : {nb_personnes_nourries:,.0f}")
print(f"≈ {nb_personnes_nourries / 1_000_000_000:.1f} milliards de personnes")

nb_personnes_nourries = 20_918_984_627_331 / 2500
```

```python
#Calcul parts alimentation Humaine, animale et pertes
#Sommes globales
total_humaine = dispo_alimentaire["Nourriture"].sum()
total_animale = dispo_alimentaire["Aliments pour animaux"].sum()
total_pertes = dispo_alimentaire["Pertes"].sum()

#Total général
total_dispo = total_humaine + total_animale + total_pertes

#Pourcentages
part_humaine = total_humaine / total_dispo * 100
part_animale = total_animale / total_dispo * 100
part_pertes = total_pertes / total_dispo * 100

#Affichage
print(f"Part alimentation humaine : {part_humaine:.1f} %")
print(f"Part alimentation animale : {part_animale:.6f} %")
print(f"Part perdue : {part_pertes:.1f} %")
```

```python
#Création d'une liste avec toutes les variables
liste_variables = dispo_pop.columns.tolist()

# Affichage de la liste
print("Liste des variables :")
for var in liste_variables:
    print(f"- {var}")
```

```python
#Création d'un dataframe avec les informations uniquement pour ces céréales
liste_cereales = [
    "Blé",
    "Maïs",
    "Millet",
    "Orge",
    "Avoine",
    "Riz (Eq Blanchi)",
    "Seigle",
    "Sorgho",
    "Céréales, Autres"
]

cereales = dispo_pop[dispo_pop["Produit"].isin(liste_cereales)]

# Affichage des premières lignes
print(cereales.shape)
cereales.head()
```

```python
#Affichage de la proportion d'alimentation animale
# Regroupement par pays
cereales_regroupees = cereales.groupby("Zone").agg({
    "Aliments pour animaux": "sum",
    "Production": "sum"
}).reset_index()

# Calcul de la proportion globale
cereales_regroupees["Proportion alimentation animale"] = cereales_regroupees.apply(
    lambda row: row["Aliments pour animaux"] / row["Production"] if row["Production"] != 0 else 0,
    axis=1
)

# Format d’affichage en pourcentage avec 6 décimales
cereales_regroupees["Proportion (%)"] = cereales_regroupees["Proportion alimentation animale"] * 100
pd.set_option("display.float_format", "{:.6f}".format)

# Formatage des colonnes numériques
cereales_regroupees["Aliments pour animaux"] = cereales_regroupees["Aliments pour animaux"].apply(lambda x: f"{x:,.0f}".replace(",", " "))
cereales_regroupees["Production"] = cereales_regroupees["Production"].apply(lambda x: f"{x:,.0f}".replace(",", " "))
cereales_regroupees["Proportion (%)"] = cereales_regroupees["Proportion (%)"].apply(lambda x: f"{x:.2f} %")

# Affichage final
print(cereales_regroupees[["Zone", "Aliments pour animaux", "Production", "Proportion (%)"]])
```

```python
# Parmi les deux usages alimentaires des principales céréales
total_humain_cereales = cereales["Nourriture"].sum()
total_animaux_cereales = cereales["Aliments pour animaux"].sum()
total_deux_usages = total_humain_cereales + total_animaux_cereales
part_humaine_cereales = 100 * total_humain_cereales / total_deux_usages
part_animale_cereales = 100 * total_animaux_cereales / total_deux_usages
print(f"Alimentation humaine : {total_humain_cereales / 1_000_000_000:,.1f} millions de tonnes ({part_humaine_cereales:.1f} %)")
print(f"Alimentation animale : {total_animaux_cereales / 1_000_000_000:,.1f} millions de tonnes ({part_animale_cereales:.1f} %)")
```

```python
#calcul du total de l'aide alimentaire par pays
# Regrouper par pays et sommer les valeurs
aide_totale_par_pays = aide_alimentaire.groupby("Zone")["Valeur"].sum().reset_index()
aide_totale_par_pays = aide_totale_par_pays.rename(columns={"Valeur": "Aide alimentaire totale (kg)"})


# Trier par ordre décroissant
aide_totale_par_pays = aide_totale_par_pays.sort_values("Aide alimentaire totale (kg)", ascending=False)
```

<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.8 - Evolution des 5 pays qui ont le plus bénéficiés de l'aide alimentaire entre 2013 et 2016</h3>
</div>

```python
#Création d'un dataframe avec la zone, l'année et l'aide alimentaire puis groupby sur zone et année 
#Filtrer les années 2013 à 2016
aide_2013_2016 = aide_alimentaire[aide_alimentaire["Année"].between(2013, 2016)]

df_evolution = aide_2013_2016[["Zone", "Année", "Valeur"]]

# 3. Grouper par pays (Zone) et année, puis sommer l’aide
df_evolution_grouped = df_evolution.groupby(["Zone", "Année"])["Valeur"].sum().reset_index()
```

```python
#Création d'une liste contenant les 5 pays qui ont le plus bénéficiées de l'aide alimentaire
top5_pays = df_evolution_grouped.groupby("Zone")["Valeur"].sum().nlargest(5).index
df_top5 = df_evolution_grouped[df_evolution_grouped["Zone"].isin(top5_pays)]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df_top5 = df_top5.copy()
df_top5["Millions de kg"] = df_top5["Valeur"] / 1_000_000

plt.figure(figsize=(10,6))
sns.lineplot(data=df_top5, x="Année", y="Millions de kg", hue="Zone", marker="o")
plt.title("Évolution de l'aide alimentaire (2013–2016) pour les 5 pays les plus aidés")
plt.ylabel("Millions de kg")
plt.xlabel("Année")
plt.xticks([2013, 2014, 2015, 2016])
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Affichage des pays avec l'aide alimentaire par année
df_evolution_grouped_sorted = df_evolution_grouped.sort_values(["Année", "Valeur"], ascending=[True, False])
df_pivot = df_evolution_grouped.pivot_table(index="Zone", columns="Année", values="Valeur", aggfunc="sum")
# Une case vide signifie que le fichier ne renseigne pas cette année
df_pivot_formatted = df_pivot.style.format("{:,.0f}", na_rep="—").highlight_max(axis=1, color="green")
df_pivot_formatted
```

```python
#création d'un dataframe avec uniquement la Thaïlande 
thailande_df = dispo_alimentaire[dispo_alimentaire["Zone"] == "Thaïlande"]
```

```python
#Calcul de la sous nutrition en Thaïlande
# Somme des kcal/personne/jour pour tous les produits
kcal_thailande = thailande_df["Disponibilité alimentaire (Kcal/personne/jour)"].sum()

seuil_min = 2300
ecart = kcal_thailande - seuil_min

if kcal_thailande < seuil_min:
    print("La Thaïlande présente un risque de sous-nutrition.")
else:
    print("La Thaïlande semble couvrir les besoins énergétiques moyens.")

print("Disponibilité totale :", kcal_thailande)



```

```python
# Filtrer la Thaïlande
thailande_df = dispo_alimentaire[dispo_alimentaire["Zone"] == "Thaïlande"].copy()

# Supprimer les cas aberrants : production = 0 mais exportations > 0
thailande_df = thailande_df[
    ~((thailande_df["Production"] == 0) & 
      (thailande_df["Exportations - Quantité"] > 0))
]

# Calculer la proportion exportée
thailande_df["Proportion exportée"] = thailande_df.apply(
    lambda row: row["Exportations - Quantité"] / row["Production"]
    if row["Production"] != 0 else 0,
    axis=1
)

# Supprimer les proportions impossibles (>100%)
thailande_df = thailande_df[thailande_df["Proportion exportée"] <= 1]

# Supprimer un cas aberrant connu
thailande_df = thailande_df[
    ~((thailande_df["Produit"] == "Crème") & 
      (thailande_df["Production"] == 1_000_000))
]

# Arrondir les colonnes numériques
thailande_df["Production"] = thailande_df["Production"].round(0).astype(int)
thailande_df["Exportations - Quantité"] = thailande_df["Exportations - Quantité"].round(0).astype(int)

# Formatage propre avec séparateur de milliers
thailande_df["Production"] = thailande_df["Production"].apply(lambda x: f"{x:,.0f}".replace(",", " "))
thailande_df["Exportations - Quantité"] = thailande_df["Exportations - Quantité"].apply(lambda x: f"{x:,.0f}".replace(",", " "))

# Pourcentage propre à 2 décimales
thailande_df["Proportion exportée (%)"] = thailande_df["Proportion exportée"].apply(
    lambda x: f"{x*100:.2f}%"
)

# Top 10 des produits les plus exportés
top_exports = thailande_df.sort_values(
    "Proportion exportée", ascending=False
)[[
    "Produit", 
    "Production", 
    "Exportations - Quantité", 
    "Proportion exportée (%)"
]].head(10)

print(top_exports)


```

```python
import matplotlib.pyplot as plt

# Sélection des 10 produits les plus exportés
top_exports = thailande_df.sort_values("Proportion exportée", ascending=False).head(10)

# Couleurs : manioc en orange, le reste en teal
colors = ["orange" if prod.lower() == "manioc" else "teal" 
          for prod in top_exports["Produit"]]

plt.figure(figsize=(10, 6))
bars = plt.barh(top_exports["Produit"], top_exports["Proportion exportée"], color=colors)

# Annotation des pourcentages
for bar, pct in zip(bars, top_exports["Proportion exportée"]):
    plt.text(
        bar.get_width() + 0.01,          # position x (juste à droite de la barre)
        bar.get_y() + bar.get_height()/2,
        f"{pct*100:.1f}%",               # format pourcentage
        va='center',
        fontsize=10
    )

plt.xlabel("Proportion exportée")
plt.title("Top 10 des produits les plus exportés en Thaïlande")
plt.gca().invert_yaxis()
plt.xlim(0, 1.1)  # pour laisser de la place aux annotations
plt.show()

```

```python
# Identification des pays d'Afrique à partir des tables de correspondance fournies
iso = pd.read_csv("Donnees p4/ISO1.csv", keep_default_na=False)
correspondance = pd.read_csv("Donnees p4/mapping_countries_FR_EN.csv", sep=";")
afrique = set(
    iso.loc[iso["Region"] == "Africa"]
    .merge(correspondance, left_on="CountryName", right_on="Country_EN")["Country_FR"]
)

# Moyennes simples par pays : disponibilité alimentaire et production
indicateurs_pays = dispo_alimentaire.groupby("Zone").agg({
    "Disponibilité alimentaire (Kcal/personne/jour)": "sum",
    "Production": "sum"
})
pays_afrique = indicateurs_pays.loc[indicateurs_pays.index.isin(afrique)]
moy_afrique = pays_afrique["Disponibilité alimentaire (Kcal/personne/jour)"].mean()
moy_monde = indicateurs_pays["Disponibilité alimentaire (Kcal/personne/jour)"].mean()
prod_moy_afrique = pays_afrique["Production"].mean() / 1_000_000_000
prod_moy_monde = indicateurs_pays["Production"].mean() / 1_000_000_000

import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].bar(["Afrique", "Monde"], [moy_afrique, moy_monde], color=["orange", "teal"])
axes[0].set_title("Disponibilité alimentaire moyenne")
axes[0].set_ylabel("kcal/personne/jour")
axes[1].bar(["Afrique", "Monde"], [prod_moy_afrique, prod_moy_monde], color=["orange", "teal"])
axes[1].set_title("Production moyenne par pays")
axes[1].set_ylabel("Millions de tonnes")
for axe, valeurs in zip(axes, ([moy_afrique, moy_monde], [prod_moy_afrique, prod_moy_monde])):
    for i, valeur in enumerate(valeurs):
        axe.text(i, valeur, f"{valeur:,.1f}", ha="center", va="bottom")
plt.tight_layout()
plt.show()
print(f"Disponibilité moyenne : Afrique {moy_afrique:.0f}, monde {moy_monde:.0f} kcal/personne/jour")
```
